## Notebook 2: Data Cleaning

```python
import pandas as pd 
import numpy as np
```

```python
file_path = "../data/processed/online_retail_II_merged.csv"

df = pd.read_csv(file_path)
```

```python
df.isnull().sum()
```

```python
df[df["Description"].isnull()].head()
```

```python
df[df["Description"].isnull()].shape
```

```python
print("Before Cleaning:", df.shape)
```

```python
df = df.dropna(subset=["Description"])
```

```python
print("After Removing Missing Description:", df.shape)
```

```python
df[df["Customer ID"].isnull()].head()
```

```python
print("Before Removing Missing Customer ID:", df.shape)
```

```python
df = df.dropna(subset=["Customer ID"])
```

```python
print("After Removing Missing Customer ID:", df.shape)
```

### Step 3: Investigate Negative Quantity

Check negative quantity values to understand whether they represent returns or invalid transactions before cleaning.

```python
df[df["Quantity"] < 0].head()
```

### Step 4: Verify Cancelled Transactions

Check whether negative quantity values correspond to cancelled invoices before removing them.

```python
df[df["Quantity"] < 0]["Invoice"].str.startswith("C").value_counts()
```

### Step 5: Remove Cancelled Transactions

Remove cancelled transactions (negative quantities) to keep only completed sales for further analysis.

```python
print("Before Removing Cancelled Transactions:")
print(df.shape)

df = df[df["Quantity"] > 0]

print("\nAfter Removing Cancelled Transactions:")
print(df.shape)
```

### Step 6: Check for Zero or Negative Prices

Inspect transactions with zero or negative prices before deciding whether to remove them.

```python
df[df["Price"] <= 0].head()
```

```python
df[df["Price"] <= 0].shape
```

### Step 7: Remove Invalid Price Values

Remove transactions with zero or negative prices to retain only valid sales records.

```python
print("Before Removing Invalid Prices:")
print(df.shape)

df = df[df["Price"] > 0]

print("\nAfter Removing Invalid Prices:")
print(df.shape)
```

### Step 8: Check for Duplicate Records

Identify duplicate records before deciding whether to remove them.

```python
df.duplicated().sum()
```

### Step 9: Inspect Duplicate Records

Examine duplicate records before removing them to ensure they are exact duplicates.

```python
df[df.duplicated()].head()
```

### Step 10: Remove Duplicate Records

Remove exact duplicate records to ensure each transaction is unique.

```python
print("Before Removing Duplicates:")
print(df.shape)

df = df.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(df.shape)
```

### Step 11: Validate Cleaned Dataset

Verify that the dataset no longer contains missing values, cancelled transactions, invalid prices, or duplicate records.

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nNegative Quantities:", (df["Quantity"] < 0).sum())

print("\nZero or Negative Prices:", (df["Price"] <= 0).sum())
```

### Step 12: Save the Cleaned Dataset

Save the cleaned dataset for use in the next stages of the project.

```python
df.to_csv("../data/processed/online_retail_II_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
