# 3. Load Input Datasets

## Objective

Load all processed datasets generated from previous notebooks. These datasets will be used to prepare dashboard-ready tables for the RetailPulse Streamlit application.

```python
# Load transaction-level dataset
transactions = pd.read_csv(
    "../data/processed/online_retail_II_feature_engineered.csv"
)

# Load customer segmentation dataset
customer_segments = pd.read_csv(
    "../data/processed/customer_segments.csv"
)

# Load customer features dataset
customer_features = pd.read_csv(
    "../data/processed/customer_features.csv"
)

# Load customer churn predictions
customer_churn = pd.read_csv(
    "../data/processed/customer_churn_predictions.csv"
)

# Load daily sales dataset
daily_sales = pd.read_csv(
    "../data/processed/daily_sales.csv"
)

# Load demand forecast dataset
forecast = pd.read_csv(
    "../data/processed/forecast.csv"
)

# Load inventory dataset
inventory = pd.read_csv(
    "../data/processed/inventory_analysis.csv"
)

print("All datasets loaded successfully!")
```

# 5. Executive Overview Dataset

## Objective

Prepare a dashboard-ready dataset containing overall business KPIs and summary metrics. This dataset will be used in the Executive Overview page of the RetailPulse Streamlit dashboard.

```python
# Calculate Executive Overview KPIs

executive_overview = pd.DataFrame({
    "TotalRevenue": [transactions["Revenue"].sum()],
    "TotalOrders": [transactions["Invoice"].nunique()],
    "TotalCustomers": [transactions["Customer ID"].nunique()],
    "TotalProducts": [transactions["StockCode"].nunique()],
    "TotalQuantitySold": [transactions["Quantity"].sum()],
    "AverageOrderValue": [
        transactions.groupby("Invoice")["Revenue"].sum().mean()
    ],
    "AverageBasketSize": [
        transactions["BasketSize"].mean()
    ],
    "CountriesServed": [
        transactions["Country"].nunique()
    ]
})

print("Executive Overview dataset created successfully!")

display(executive_overview)
```

# 6. Sales Analytics Dataset

## Objective

Prepare a dashboard-ready dataset containing daily sales performance. This dataset will support sales trend analysis, revenue visualization, and business performance monitoring in the Streamlit dashboard.

```python
# Prepare Sales Analytics dataset

sales_analytics = daily_sales.copy()

print("Sales Analytics dataset created successfully!")

print(f"Dataset Shape: {sales_analytics.shape}")

display(sales_analytics.head())
```

# 7. Customer Segmentation Dataset

## Objective

Prepare a dashboard-ready dataset containing customer segmentation information generated in Notebook 05. This dataset will support customer segment analysis and visualization in the Streamlit dashboard.

# 9. Demand Forecasting Dataset

## Objective

Prepare a dashboard-ready dataset containing historical sales and forecasted demand generated in Notebook 06. This dataset will support demand forecasting visualization and trend analysis in the RetailPulse Streamlit dashboard.

```python
# Prepare dashboard-ready Demand Forecasting dataset

demand_forecasting = forecast[
    [
        "ds",
        "yhat",
        "yhat_lower",
        "yhat_upper"
    ]
].copy()

print("Demand Forecasting dataset created successfully!")

print(f"Dataset Shape: {demand_forecasting.shape}")

display(demand_forecasting.head())
```

# 10. Customer Churn Dataset

## Objective

Prepare a dashboard-ready dataset containing customer churn predictions and churn probability scores generated in Notebook 08. This dataset will support customer churn analysis, risk identification, and retention strategy visualization in the RetailPulse Streamlit dashboard.

```python
# Prepare Customer Churn dataset

customer_churn_dashboard = customer_churn.copy()

print("Customer Churn dataset created successfully!")

print(f"Dataset Shape: {customer_churn_dashboard.shape}")

display(customer_churn_dashboard.head())
```

# 11. Inventory Optimization Dataset

## Objective

Prepare a dashboard-ready dataset containing product inventory metrics and ABC classification generated in Notebook 09. This dataset will support inventory monitoring, stock prioritization, and inventory optimization analysis in the RetailPulse Streamlit dashboard.

```python
# Prepare dashboard-ready Inventory Optimization dataset

inventory_dashboard = inventory[
    [
        "StockCode",
        "Description",
        "TotalQuantity",
        "TotalRevenue",
        "TotalOrders",
        "AveragePrice",
        "ABC_Class",
        "MovementCategory"
    ]
].copy()

print("Inventory Optimization dataset created successfully!")

print(f"Dataset Shape: {inventory_dashboard.shape}")

display(inventory_dashboard.head())
```

# 12. Product Analytics Dataset

## Objective

Prepare a dashboard-ready dataset containing product performance metrics for sales analysis, revenue contribution, and product ranking. This dataset will support product analytics visualization in the RetailPulse Streamlit dashboard.

```python
# Prepare Product Analytics dataset

product_analytics = inventory_dashboard[
    [
        "StockCode",
        "Description",
        "TotalQuantity",
        "TotalRevenue",
        "TotalOrders",
        "AveragePrice"
    ]
].copy()

# Sort by revenue (highest first)
product_analytics = product_analytics.sort_values(
    by="TotalRevenue",
    ascending=False
).reset_index(drop=True)

print("Product Analytics dataset created successfully!")

print(f"Dataset Shape: {product_analytics.shape}")

display(product_analytics.head())
```

# 13. Country Analytics Dataset

## Objective

Prepare a dashboard-ready dataset containing country-wise sales performance and customer distribution. This dataset will support geographical sales analysis and country performance visualization in the RetailPulse Streamlit dashboard.

```python
# Prepare Country Analytics dataset

country_analytics = (
    transactions.groupby("Country")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        TotalOrders=("Invoice", "nunique"),
        TotalCustomers=("Customer ID", "nunique"),
        TotalQuantity=("Quantity", "sum")
    )
    .reset_index()
)

# Calculate Average Revenue per Customer
country_analytics["AverageRevenuePerCustomer"] = (
    country_analytics["TotalRevenue"] /
    country_analytics["TotalCustomers"]
).round(2)

# Sort by Total Revenue
country_analytics = country_analytics.sort_values(
    by="TotalRevenue",
    ascending=False
).reset_index(drop=True)

print("Country Analytics dataset created successfully!")

print(f"Dataset Shape: {country_analytics.shape}")

display(country_analytics.head())
```

# 14. Business Insights Dataset

## Objective

Prepare a dashboard-ready summary dataset containing high-level business metrics and insights derived from previous analyses. This dataset will support executive decision-making and business performance monitoring in the RetailPulse Streamlit dashboard.

```python
# Prepare Business Insights dataset

business_insights = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Customers",
        "Total Orders",
        "Countries Served",
        "Products",
        "Predicted Churn Customers",
        "Fast Moving Products",
        "Category A Products"
    ],
    "Value": [
        transactions["Revenue"].sum(),
        transactions["Customer ID"].nunique(),
        transactions["Invoice"].nunique(),
        transactions["Country"].nunique(),
        transactions["StockCode"].nunique(),
        customer_churn_dashboard["PredictedChurn"].sum(),
        (inventory_dashboard["MovementCategory"] == "Fast Moving").sum(),
        (inventory_dashboard["ABC_Class"] == "A").sum()
    ]
})

print("Business Insights dataset created successfully!")

display(business_insights)
```
